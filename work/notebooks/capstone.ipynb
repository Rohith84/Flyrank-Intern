# Research Question

How can search performance metrics be combined to identify content pages that should be prioritized for refresh and optimization?

## Decision Supported

This project helps SEO and content teams decide which pages should be refreshed first based on search visibility and user engagement metrics.

# Data

**Data Sources:**

Missing GA4 values were replaced with zero, while missing average search position values were replaced with 100. No client names, domains, URLs, or private information were included.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

# Methodology

## Assumptions

- Missing GA4 values indicate no recorded activity and were replaced with zero.
- Missing average search position values were replaced with 100.
- The Opportunity Score uses manually assigned feature weights.

## Features

- Google Search Console Impressions
- Click Through Rate (CTR)
- Average Search Position
- Engagement Per Session

## Baseline

The baseline approach ranks pages using impressions only.

## Proposed Method

A weighted Opportunity Score combines impressions, CTR, ranking position, and engagement to prioritize pages for optimization.

## Validation

The generated rankings were reviewed as decision-support recommendations rather than predictive outputs.

## Leakage Checks

Only current search and engagement metrics were used. No future information was included when generating Opportunity Scores.

## 5. Limitations

*What this work cannot claim.*

# Limitations

This project uses a rule-based scoring framework rather than a supervised machine learning model.

The Opportunity Score is intended for decision support only.

The results should not be interpreted as predictions of Google's ranking algorithm or guarantees of future search performance.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

# Ranked Recommendations

- Improve titles and meta descriptions for pages with low CTR.
- Refresh pages with poor average ranking.
- Prioritize pages with high impressions.
