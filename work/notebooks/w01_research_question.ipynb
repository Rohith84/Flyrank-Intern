## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## Research Question

### Decision

Which content pages should be prioritized for content refresh based on search visibility and engagement metrics?

### Action

SEO and content teams can refresh high-priority pages by improving titles, meta descriptions, internal links, and page content.

### Unit of Analysis

Each row represents the daily performance of one content page.

### Cost of a Wrong Recommendation

If low-priority pages are selected instead of high-opportunity pages, time and SEO effort may be wasted while important pages remain unoptimized.

### Why Data Helps

Search impressions, clicks, ranking position, and engagement provide measurable signals that help prioritize pages more objectively than manual review.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
from google.colab import userdata
from huggingface_hub import login
import duckdb

HF_TOKEN = userdata.get("HF_TOKEN")
login(HF_TOKEN)

con = duckdb.connect()

con.execute(f"""
CREATE OR REPLACE SECRET hf_secret (
TYPE HUGGINGFACE,
TOKEN '{HF_TOKEN}'
);
""")

query = """
SELECT *
FROM read_parquet(
'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance_sample.parquet'
)
LIMIT 50000
"""

sample = con.execute(query).df()

print("Sample Size:", len(sample))
print("Average Impressions:", round(sample["gsc_impressions"].mean(),2))
print("Average Clicks:", round(sample["gsc_clicks"].mean(),2))
print("Missing GA4 Rows:", sample["ga4_pageviews"].isnull().sum())
```
