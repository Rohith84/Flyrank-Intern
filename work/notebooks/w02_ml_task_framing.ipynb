## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

## My Lane as an ML Task

**Selected Lane:** Refresh / Content Opportunity Scoring

**ML Task Type:** Scoring / Ranking

This project is a scoring and ranking task because the objective is to prioritize content pages according to their optimization opportunity. Instead of predicting a class label, the system generates an Opportunity Score that ranks pages from highest to lowest priority for content refresh.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## Target or Proxy

The project does not use a supervised target label. Instead, a proxy target called the **Opportunity Score** is calculated using Google Search Console and Google Analytics metrics.

The score combines impressions, click-through rate (CTR), average ranking position, and engagement into a single value. This proxy score is then used to rank content pages for review.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## Success Metric

The success of this project is measured by the quality of the generated ranking.

A useful ranking should prioritize pages that have high visibility but low performance, allowing content teams to focus their optimization efforts on pages with the greatest opportunity for improvement.

The generated rankings should provide meaningful decision-support recommendations rather than predictions.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
from google.colab import userdata
from huggingface_hub import login
import duckdb

HF_TOKEN = userdata.get("HF_TOKEN")
login(HF_TOKEN)

con = duckdb.connect()

con.execute(f"""
CREATE OR REPLACE SECRET hf_secret (
TYPE HUGGINGFACE,
TOKEN '{HF_TOKEN}'
);
""")

query = """
SELECT *
FROM read_parquet(
'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance_sample.parquet'
)
LIMIT 5
"""

sample = con.execute(query).df()

sample[[
    "content_hash_id",
    "report_date",
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position"
]]
```

### Unit of Analysis

Each row represents the daily search performance of a single content page.

The dataset includes search visibility, ranking, click, engagement, and AI referral metrics that are used to calculate the Opportunity Score.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML Beats a Fixed Rule

A simple rule such as selecting pages with the highest impressions cannot capture the full relationship between visibility, click-through rate, ranking position, and user engagement.

A scoring approach combines multiple search signals into a single ranking, making it easier to prioritize pages with the greatest optimization opportunity.

This project is intended to support content decisions using observed search data rather than relying on a single threshold or manually defined rule.
